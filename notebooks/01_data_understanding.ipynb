```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
```

```python
df = pd.read_csv('../data/raw/creditcard.csv')
print("Dataset loaded successfully")
print(f"Shape: {df.shape}")
```

```python
df = df.drop_duplicates()

print("New shape:", df.shape)
```

```python
df.duplicated().sum()
df["Class"].value_counts(normalize=True) * 100
```

```python
sns.countplot(x='Class', data=df)
plt.title("Class Distribution (0=Normal, 1=Fraud)")
plt.show()
```

How does Amount behave for the different classes?

```python
sns.boxplot(x='Class', y='Amount', data=df)
plt.title("Amount vs Class")
plt.show()
```

Fraud transactions tend to be lower in amount compared to normal transactions

```python
sns.histplot(df['Time'], bins=50)
plt.title("Distribution of Transaction Time")
plt.show()
```

```python
sns.boxplot(x='Class', y='Time', data=df)
plt.title("Time vs Class")
plt.show()
```

```python
sns.kdeplot(data=df[df['Class']==0]['Time'], label='Normal', fill=True)
sns.kdeplot(data=df[df['Class']==1]['Time'], label='Fraud', fill=True)
plt.title("Time Distribution: Fraud vs Normal")
plt.legend()
plt.show()
```

```python
corr = df.corr()

# Focus on correlation with Class
corr_with_class = corr['Class'].sort_values(ascending=False)

print(corr_with_class)
```

```python
top_features = corr_with_class.index[1:6]   # top 5 (excluding Class itself)
bottom_features = corr_with_class.index[-5:]  # most negative
```

```python
for feature in top_features:
    sns.kdeplot(df[df['Class']==0][feature], label='Normal', fill=True)
    sns.kdeplot(df[df['Class']==1][feature], label='Fraud', fill=True)
    plt.title(f"{feature} Distribution")
    plt.legend()
    plt.show()
```
