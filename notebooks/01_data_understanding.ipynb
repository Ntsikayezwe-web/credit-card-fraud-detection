Amounts Distributions 

```python
fig, axes = plt.subplots(1,3, figsize=(15,20))

ax1 = axes[0]
df['Amount'].hist(bins=100, ax=ax1)
ax1.set_title('Transaction Amount Distribution')
ax1.set_xlabel('Amount')
ax1.set_ylabel('Frequency')
ax1.set_yscale('log') #rescale so that we see clearly for the bin that apears the least

ax2 = axes[1]

#rescale amounts using log to shrink the gap between minimum vlaues and maximum values
amount_log = np.log10(df['Amount'] + 1) #add 1 to avoid 0 dollar transaction
normal_amount_log = np.log10(df[df['Class']==0]['Amount'] + 1)
fraud_amount_log = np.log10(df[df['Class']==1]['Amount'] + 1)

ax2.hist([normal_amount_log, fraud_amount_log], bins=50, label=['Normal','Fraud'], color=['green', 'red'])
ax2.set_title('Amount Distribution by Class')
ax2.set_xlabel('Amount')
ax2.set_ylabel('Frequency')
ax2.legend()


#By looking at where these two curves peak and where they overlap, see if Fraudsters tend to "hang out" at different price points than Normal users.
ax3 = axes[2]
df[df['Class']==0]['Amount'].plot.density(ax=ax3, label='Normal', color='green')
df[df['Class']==1]['Amount'].plot.density(ax=ax3, label='Fraud', color='red')
ax3.set_title('Amount Density Distribution')
ax3.set_xlabel('Amount')
ax3.set_xlim(0,5000) # from the first plot, this is where most amount were in
ax3.legend()

plt.tight_layout()
plt.show()
```

Time Analysis

```python
df['Hour']= df['Time'] / 3600

fig, axes = plt.subplots(1,2, figsize=(15,6))

#transactions by class
ax1 = axes[0]
normal_time = df[df['Class']==0]['Hour']
fraud_time = df[df['Class']==1]['Hour']
ax1.hist([normal_time,fraud_time], bins=58, label=['Normal','Fraud'], color=['green','red'],density=True)
ax1.set_title('Time Distribution by Class (Normalized)')
ax1.set_xlabel('Hours from Start')
ax1.set_ylabel('Density')
ax1.legend()


# 4. Scatter plot: Time vs Amount
ax2 = axes[1]

# Sample normal transactions for visibility
normal_sample = df[df['Class']==0].sample(1000, random_state=42)
fraud_all = df[df['Class']==1]

ax2.scatter(normal_sample['Hour'], normal_sample['Amount'], 
           alpha=0.5, s=10, label='Normal (sample)', color='green')
ax2.scatter(fraud_all['Hour'], fraud_all['Amount'], 
           alpha=0.7, s=20, label='Fraud (all)', color='red')
ax2.set_title('Time vs Amount Pattern', fontsize=12, fontweight='bold')
ax2.set_xlabel('Hours from Start')
ax2.set_ylabel('Amount ($)')
ax2.set_yscale('log')
ax2.legend()
```

Calculate the effect for the pca transformed features

```python
pca_features = [col for col in df.columns if col.startswith('V')]


def calculate_effect_size(feature):
    """Calculate Cohen's d effect size for a feature"""
    normal_data = df[df['Class'] == 0][feature]
    fraud_data = df[df['Class'] == 1][feature]
    
    mean_diff = abs(normal_data.mean() - fraud_data.mean())
    pooled_std = np.sqrt((normal_data.std()**2 + fraud_data.std()**2) / 2)
    
    return mean_diff / pooled_std if pooled_std > 0 else 0

# Calculate effect sizes for all PCA features
effect_sizes = []
for feature in pca_features:
    effect_size = calculate_effect_size(feature)
    effect_sizes.append({
        'Feature': feature,
        'Effect_Size': effect_size,
        'Normal_Mean': df[df['Class'] == 0][feature].mean(),
        'Fraud_Mean': df[df['Class'] == 1][feature].mean()
    })

# Create DataFrame and sort by effect size
effect_df = pd.DataFrame(effect_sizes).sort_values('Effect_Size', ascending=False)

print(effect_df.head(12))
```
